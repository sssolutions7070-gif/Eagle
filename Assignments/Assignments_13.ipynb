```python

""""
Write a program to read a file and find the longest word present in it.
"""

maxlen = 0
longest = ""

with open("python.txt", "r") as f:
    for i in f:
        words = i.split()

        for word in words:
            if len(word) > maxlen:
                maxlen = len(word)
                longest = word

print("Longest word:", longest)
print("Length:", maxlen)
```

```python
"""
Write a program to read a file containing numbers and calculate their sum, average,
maximum, and minimum.
"""


numbers = []

with open("num.txt", "r") as f:
    for i in f:
        numbers.append(int(i))

total = sum(numbers)
average = total / len(numbers)
maximum = max(numbers)
minimum = min(numbers)

print("total", total)
print("average", average)
print("maximum", maximum)
print("minimum", minimum)
```
