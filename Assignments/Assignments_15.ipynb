```python
"""
1. Student Class – __str__()
Create a Student class with name, roll_no, and marks. Implement __str__() so that
printing the

"""

class student:
    def __init__(self,name,roll_no,marks):
        self.name=name
        self.roll_no=roll_no
        self.marks=marks

    def __str__(self):
        print("Name:",self.name)
        print("Roll No:",self.roll_no)
        print("Marks:",self.marks)


student1=student("satej",12,89)
student1.__str__()

```

```python
"""
2. Product Class – __eq__()
Create a Product class with name and price. Implement __eq__() to check whether two
products have the same price.

"""
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price

    def __eq__(self,other):
        return self.price==other.price

product1=product("mobile",34000)
product2=product("iphone",34000)
product3=product("vivo",12000)


print(product1==product2)
print(product2==product3)
print(product1==product3)
    


```

```python
"""
3. Shopping Cart – __len__()
Create a ShoppingCart class containing a list of products. Implement __len__() so that
len(cart) returns the number of products in the cart.

"""

class ShoppingCart:
    def __init__(self,products):
        self.products=products

    def __len__(self):
        return len(self.products)

cart=ShoppingCart(["bags","pen","notebook","book"])

print(len(cart))
```

```python
"""
4. Bank Account – __str__()
Create a BankAccount class with account holder name, account number, and balance.
Implement __str__() to display account information when the object is printed.

"""

class BankAccount:
    def  __init__(self,acc_name,acc_number,acc_balance):
        self.acc_name=acc_name
        self.acc_number=acc_number
        self.acc_balance=acc_balance


    def __str__(self):
        print("Account Holder Name:",self.acc_name)
        print("Account Number:",self.acc_name)
        print("Account Balance:",self.acc_balance)


bank1=BankAccount("Satej Rajeshirke",12345678,4500)
bank2=BankAccount("Jay",56578988,9500)
bank3=BankAccount("Eshwar",67676,10500)

bank1.__str__()
bank2.__str__()
bank3.__str__()
```

```python
"""
5. Number Class – __add__()
Create a Number class that stores one number. Implement __add__() so that two
objects can be added

"""

class Number:
    def __init__(self,numbers):
        self.numbers=numbers

    def __add__(self,other):
        return self.numbers + other.numbers

op1=Number(10)
op2=Number(56)

print(op1+op2)
```

```python
"""
6. Create a Number class and implement:
__add__()
__sub__()
__mul__()
__truediv__()
__mod__()

"""

class Number:
    def __init__(self,num1,num2):
        self.num1=num1
        self.num2=num2

    def __add__(self):
        return self.num1+self.num2


    def __sub__(self):
        return self.num1-self.num2

    def __mul__(self):
        return self.num1*self.num2

    def __truedivs__(self):
        return self.num1/self.num2

    def __mod__(self):
        return self.num1%self.num2


ob1=Number(10,20)
print(ob1.__add__())
print(ob1.__sub__())
print(ob1.__mul__())
print(ob1.__truedivs__())
print(ob1.__mod__())
```
